```python
import pandas as pd
import os

# Updated target list based on your specific requirements
TARGETS = [
    'commute', 'jbmth', 'home7', 'traccess', 
    'nbrsnc', 'jbisic8_dv', 'jbsoc10_dv'
]

def diagnostic_intake(wave_letter):
    file_path = f"../data/ukhls/raw/{wave_letter}_indresp.tab"
    if not os.path.exists(file_path):
        return print(f"File {wave_letter} not found.")

    # Load just the first row to see ALL column names
    all_headers = pd.read_csv(file_path, sep='\t', nrows=0).columns.tolist()
    
    found_cols = []
    for t in TARGETS:
        # Check for w_variable (e.g., o_commute)
        prefixed = f"{wave_letter}_{t}"
        # Check for raw variable (e.g., commute)
        raw = t
        
        if prefixed in all_headers:
            found_cols.append(prefixed)
        elif raw in all_headers:
            found_cols.append(raw)
        else:
            # Last ditch: search for any column that CONTAINS the name
            fuzzy_match = [c for c in all_headers if t in c]
            if fuzzy_match:
                print(f"Found fuzzy match for {t}: {fuzzy_match[0]}")
                found_cols.append(fuzzy_match[0])
            else:
                print(f"!!! {t} is TRULY MISSING in Wave {wave_letter}")

    return found_cols

# Run diagnostic for Wave O
diagnostic_intake('l')
```
