```python
"""0 — Run pipeline (steps 1 → 9).

Runs each notebook in order via jupyter nbconvert --execute.
Start Jupyter from the repo root or data_pipeline/.

Steps:
  1  pickle_sipher           — pickle SIPHER population data
  2  pickle_ukhls_waves      — pickle UKHLS wave data (merges hhresp if present)
  3  backfill_ukhls_waves    — backfill missing values across waves
  4  feature_eng_ukhls       — feature engineering + OHE
  5  add_nl_strings          — generate natural language profiles
  6  add_vector_embedding    — embed profiles via OpenAI API
  7  add_sipher_pop_row_counts — append n_sipher_rows population weights
  8  cluster_embeddings_k_means — K-Means on embeddings → embedding_k_means_cluster
  9  cluster_values_k_means  — K-Means on feature vars → values_k_means_cluster
"""

import importlib
import os
import subprocess
import sys
from pathlib import Path


def _pipeline_dir() -> Path:
    here = Path.cwd().resolve()
    if (here / "config_variables.py").is_file():
        return here
    if (here / "data_pipeline" / "config_variables.py").is_file():
        return here / "data_pipeline"
    raise FileNotFoundError(
        "Start Jupyter from the repo root or from data_pipeline/ so config_variables.py can be found."
    )


PIPELINE_DIR = _pipeline_dir()
os.chdir(PIPELINE_DIR)
sys.path.insert(0, str(PIPELINE_DIR))

import config_variables
importlib.reload(config_variables)

print("Using:", sys.executable)
print("DATA_FOLDER:", config_variables.DATA_FOLDER)
print("cwd:", Path.cwd())

NOTEBOOKS = [
    "1_pickle_sipher.ipynb",
    "2_pickle_ukhls_waves.ipynb",
    "3_backfill_ukhls_waves.ipynb",
    "4_feature_eng_ukhls.ipynb",
    "5_add_nl_strings.ipynb",
    "6_add_vector_embedding.ipynb",
    "7_add_sipher_pop_row_counts.ipynb",
    "8_cluster_embeddings_k_means.ipynb",
    "9_cluster_values_k_means.ipynb",
]

for nb in NOTEBOOKS:
    print("\n" + "=" * 46)
    print("  EXECUTE ", nb)
    print("=" * 46)
    out_tmp = f"{nb}.executed.tmp.ipynb"
    subprocess.run(
        [
            sys.executable,
            "-m",
            "jupyter",
            "nbconvert",
            "--to",
            "notebook",
            "--execute",
            nb,
            "--output",
            out_tmp,
        ],
        check=True,
        cwd=PIPELINE_DIR,
    )
    tmp_path = PIPELINE_DIR / out_tmp
    if tmp_path.exists():
        tmp_path.unlink()
    print("  OK:", nb)

print("\nPipeline complete: steps 1–9 finished.")
```
