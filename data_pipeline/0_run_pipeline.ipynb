```python
"""0 — Run pipeline (main path: 1 → 2 → 3a → 4a → 5a → 6a → 7 → 8 → 9).

Runs each notebook in order via jupyter nbconvert --execute.
Start Jupyter from the repo root or data_pipeline/. Step 6a (synthetic population) can take a long time on full data/.
Optional validation notebooks: 3b, 4b, 5b, 6b. Optional later: 10_label_local_level_clusters, 11_label_national_level_clusters, 12_generate_portraits, 13_nl_profiles, 14_embed_profiles.
"""

import importlib.util
import os
import subprocess
import sys
from pathlib import Path


def _pipeline_dir() -> Path:
    here = Path.cwd().resolve()
    if (here / "config_paths.py").is_file():
        return here
    if (here / "data_pipeline" / "config_paths.py").is_file():
        return here / "data_pipeline"
    raise FileNotFoundError(
        "Start Jupyter from the repo root or from data_pipeline/ so config_paths.py can be found."
    )


PIPELINE_DIR = _pipeline_dir()
os.chdir(PIPELINE_DIR)
sys.path.insert(0, str(PIPELINE_DIR))

# Load config_paths from disk every time — avoids a stale cached `config_paths`
# in a long-lived Jupyter kernel after new flags (e.g. USE_FOUR_LA_SUBSET) are added.
_cfg_path = PIPELINE_DIR / "config_paths.py"
_spec = importlib.util.spec_from_file_location("_archetypes_config_paths", _cfg_path)
cp = importlib.util.module_from_spec(_spec)
_spec.loader.exec_module(cp)

print("Using:", sys.executable)
print("DATA_FOLDER:", cp.DATA_FOLDER)
print("cwd:", Path.cwd())
if cp.USE_FOUR_LA_SUBSET:
    print("FOUR-LA SUBSET: Newham, Tower Hamlets, Islington, Hounslow only")

NOTEBOOKS = [
    "1_pickle_sipher.ipynb",
    "2_pickle_ukhls_waves.ipynb",
    "3a_backfill_ukhls_waves.ipynb",
    "4a_feature_eng_ukhls.ipynb",
    "5a_derive_variables.ipynb",
    "6a_synthetic_population.ipynb",
    "7_cluster_local_level.ipynb",
    "8_cluster_national_level.ipynb",
    "9_group_averages.ipynb",
]

for nb in NOTEBOOKS:
    print("\n" + "=" * 46)
    print("  EXECUTE ", nb)
    print("=" * 46)
    out_tmp = f"{nb}.executed.tmp.ipynb"
    subprocess.run(
        [
            sys.executable,
            "-m",
            "jupyter",
            "nbconvert",
            "--to",
            "notebook",
            "--execute",
            nb,
            "--output",
            out_tmp,
        ],
        check=True,
        cwd=PIPELINE_DIR,
    )
    tmp_path = PIPELINE_DIR / out_tmp
    if tmp_path.exists():
        tmp_path.unlink()
    print("  OK:", nb)

print("\nPipeline steps 1–9 finished.")
```
