```python
# Check for NaNs in original UKHLS .tab files (waves k-o)
import os
ukhls_tab_dir = '../data/0_raw/ukhls/'
waves = ['a','k', 'l', 'm', 'n', 'o']
for wave in waves:
    tab_path = os.path.join(ukhls_tab_dir, f'{wave}_indresp.tab')
    if not os.path.exists(tab_path):
        print(f"File not found: {tab_path}")
        continue
    df_tab = pd.read_csv(tab_path, sep='\t', low_memory=False)
    nans = df_tab.isna().sum()
    total = len(df_tab)
    nan_cols = nans[nans > 0]
    if not nan_cols.empty:
        print(f'\nNaN columns in {wave}_indresp.tab:')
        for col, n_nan in nan_cols.items():
            print(f'  {col}: {n_nan:,} NaNs ({100 * n_nan / total:.2f}%)')
    else:
        print(f'No NaNs found in {wave}_indresp.tab')
```

```python
# Check for NaNs in all intermediate UKHLS pickle files
import os
import pandas as pd
import re

pickle_dirs = [
    '../data/2_pickle_ukhls_waves',
    '../data/3_backfill_ukhls_waves',
    '../data/4_feature_eng_ukhls',
]
ukhls_wave_pattern = re.compile(r'^[a-o]_[a-zA-Z0-9_]+$')

for pkl_dir in pickle_dirs:
    if not os.path.exists(pkl_dir):
        print(f"Directory not found: {pkl_dir}")
        continue
    for fname in os.listdir(pkl_dir):
        if not fname.endswith('.pkl'):
            continue
        fpath = os.path.join(pkl_dir, fname)
        try:
            df = pd.read_pickle(fpath)
        except Exception as e:
            print(f"Failed to load {fpath}: {e}")
            continue
        ukhls_cols = [col for col in df.columns if ukhls_wave_pattern.match(col)]
        nan_report = []
        for col in ukhls_cols:
            n_nan = df[col].isna().sum()
            if n_nan > 0:
                nan_report.append((col, n_nan, 100 * n_nan / len(df)))
        print(f"\nFile: {fpath}")
        if nan_report:
            print("  NaN columns:")
            for col, n_nan, pct in sorted(nan_report, key=lambda x: -x[1]):
                print(f"    {col}: {n_nan:,} NaNs ({pct:.2f}%)")
        else:
            print("  No NaNs found in any UKHLS wave columns.")
```
